````python
import spacy

nlp = spacy.blank("en")

nlp.add_pipe(
    "transformer",
    config={
        "model": {
            "@architectures": "spacy-transformers.TransformerModel.v3",
            "name": "emilyalsentzer/Bio_ClinicalBERT",
        }
    },
)

print("Transformer loaded successfully!")
````
