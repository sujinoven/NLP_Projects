```python
# Part 3: Data Preparation

# Create a copy of the original dataset
data = df.copy()

# Remove the unnecessary index column
data.drop(columns=["Unnamed: 0"], errors="ignore", inplace=True)

# Remove duplicate rows
data.drop_duplicates(inplace=True)

# Remove records where the product name is missing
data.dropna(subset=["name"], inplace=True)

# Fill missing category values
data["main_category"] = data["main_category"].fillna("unknown")
data["sub_category"] = data["sub_category"].fillna("unknown")

# Clean the text columns
text_columns = ["name", "main_category", "sub_category"]

for column in text_columns:
    data[column] = data[column].astype(str).str.lower().str.strip()

# Convert rating columns into numbers
data["ratings"] = pd.to_numeric(data["ratings"], errors="coerce")

data["no_of_ratings"] = pd.to_numeric(
    data["no_of_ratings"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
)

# Remove currency symbols and commas from price columns
price_columns = ["discount_price", "actual_price"]

for column in price_columns:
    data[column] = pd.to_numeric(
        data[column].astype(str).str.replace(r"[₹,]", "", regex=True),
        errors="coerce"
    )

# Create one text column for the NLP similarity model
data["combined_text"] = (
    data["name"] + " " +
    data["main_category"] + " " +
    data["sub_category"]
)

# Reset the row numbers
data.reset_index(drop=True, inplace=True)

# Display the prepared dataset
print("Data preparation completed.")
print("Dataset shape:", data.shape)
print("\nMissing values:")
print(data.isnull().sum())

display(data.head())
```

```python
# Part 4: Text Preprocessing

import re
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Download the required NLTK resources
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")

# Create the stop-word list and lemmatizer
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

# Function to clean product text
def preprocess_text(text):
    # Convert text to lowercase
    text = str(text).lower()

    # Remove punctuation, numbers and special characters
    text = re.sub(r"[^a-z\s]", " ", text)

    # Split the sentence into individual words
    words = text.split()

    # Remove stop words and convert words to their base forms
    words = [
        lemmatizer.lemmatize(word)
        for word in words
        if word not in stop_words
    ]

    # Join the words back into a sentence
    return " ".join(words)

# Apply preprocessing to the combined product text
data["clean_text"] = data["combined_text"].apply(preprocess_text)

# Remove rows that contain no usable text
data = data[data["clean_text"].str.strip() != ""]

# Reset row numbers
data.reset_index(drop=True, inplace=True)

# Compare the original and cleaned text
print("Text preprocessing completed.")
display(data[["combined_text", "clean_text"]].head(10))
```

```python
# Part 5: TF-IDF Model and Cosine Similarity Search System

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# ------------------------------------------
# 1. Build the TF-IDF model
# ------------------------------------------

tfidf_model = TfidfVectorizer(
    max_features=20000,
    ngram_range=(1, 2),
    min_df=2
)

# Train the model and transform the cleaned text
tfidf_matrix = tfidf_model.fit_transform(data["clean_text"])

print("TF-IDF model built successfully.")
print("TF-IDF matrix shape:", tfidf_matrix.shape)


# ------------------------------------------
# 2. Create the recommendation function
# ------------------------------------------

def recommend_products(product_name, number_of_recommendations=5):

    # Search for the product name
    matching_products = data[
        data["name"].str.contains(
            product_name,
            case=False,
            na=False,
            regex=False
        )
    ]

    # Check whether the product exists
    if matching_products.empty:
        print("Product not found. Please try another product name.")
        return

    # Select the first matching product
    product_index = matching_products.index[0]
    selected_product = data.loc[product_index, "name"]

    print(f"\nSelected product: {selected_product}")

    # Calculate similarity with all other products
    similarity_scores = cosine_similarity(
        tfidf_matrix[product_index],
        tfidf_matrix
    ).flatten()

    # Sort products from most similar to least similar
    similar_indices = similarity_scores.argsort()[::-1]

    # Remove the selected product itself
    similar_indices = [
        index for index in similar_indices
        if index != product_index
    ]

    # Keep only the requested number of recommendations
    similar_indices = similar_indices[:number_of_recommendations]

    # Select useful result columns
    result_columns = [
        "name",
        "main_category",
        "sub_category",
        "ratings",
        "discount_price"
    ]

    recommendations = data.loc[
        similar_indices,
        result_columns
    ].copy()

    # Add similarity scores
    recommendations["similarity_score"] = similarity_scores[
        similar_indices
    ]

    # Reset row numbers
    recommendations.reset_index(drop=True, inplace=True)
    recommendations.index = recommendations.index + 1

    return recommendations


# ------------------------------------------
# 3. Test the recommendation system
# ------------------------------------------

search_product = input("\nEnter a product name: ")

results = recommend_products(
    search_product,
    number_of_recommendations=5
)

if results is not None:
    display(results)
```

```python
# Part 6: Word2Vec Model and Cosine Similarity Search System

!pip install -q gensim

import numpy as np

from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity

# ------------------------------------------
# 1. Convert cleaned text into word lists
# ------------------------------------------

sentences = data["clean_text"].apply(lambda text: text.split()).tolist()

# ------------------------------------------
# 2. Train the Word2Vec model
# ------------------------------------------

word2vec_model = Word2Vec(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=2,
    workers=4,
    epochs=10,
    seed=42
)

print("Word2Vec model trained successfully.")
print("Vocabulary size:", len(word2vec_model.wv))

# ------------------------------------------
# 3. Create one vector for each product
# ------------------------------------------

def create_product_vector(text):
    words = text.split()

    # Keep only words learned by Word2Vec
    word_vectors = [
        word2vec_model.wv[word]
        for word in words
        if word in word2vec_model.wv
    ]

    # Return the average word vector
    if word_vectors:
        return np.mean(word_vectors, axis=0)

    # Return a zero vector if no words were found
    return np.zeros(word2vec_model.vector_size)


product_vectors = np.array(
    data["clean_text"].apply(create_product_vector).tolist()
)

print("Product-vector matrix shape:", product_vectors.shape)

# ------------------------------------------
# 4. Create the recommendation function
# ------------------------------------------

def recommend_products_word2vec(product_name, number_of_recommendations=5):

    # Find products containing the search text
    matching_products = data[
        data["name"].str.contains(
            product_name,
            case=False,
            na=False,
            regex=False
        )
    ]

    if matching_products.empty:
        print("Product not found. Please try another product name.")
        return

    # Select the first matching product
    product_index = matching_products.index[0]
    selected_product = data.loc[product_index, "name"]

    print(f"\nSelected product: {selected_product}")

    # Calculate cosine similarity
    similarity_scores = cosine_similarity(
        product_vectors[product_index].reshape(1, -1),
        product_vectors
    ).flatten()

    # Sort products by similarity
    similar_indices = similarity_scores.argsort()[::-1]

    # Exclude the selected product
    similar_indices = [
        index for index in similar_indices
        if index != product_index
    ][:number_of_recommendations]

    # Prepare the results
    result_columns = [
        "name",
        "main_category",
        "sub_category",
        "ratings",
        "discount_price"
    ]

    recommendations = data.loc[
        similar_indices,
        result_columns
    ].copy()

    recommendations["similarity_score"] = similarity_scores[
        similar_indices
    ]

    recommendations.reset_index(drop=True, inplace=True)
    recommendations.index = recommendations.index + 1

    return recommendations

# ------------------------------------------
# 5. Test the recommendation system
# ------------------------------------------

search_product = input("\nEnter a product name: ")

results = recommend_products_word2vec(
    search_product,
    number_of_recommendations=5
)

if results is not None:
    display(results)
```

```python
# Part 7: FastText Model and Cosine Similarity Search System

!pip install -q gensim

import numpy as np

from gensim.models import FastText
from sklearn.metrics.pairwise import cosine_similarity

# ------------------------------------------
# 1. Convert cleaned text into word lists
# ------------------------------------------

sentences = data["clean_text"].apply(lambda text: text.split()).tolist()

# ------------------------------------------
# 2. Train the FastText model
# ------------------------------------------

fasttext_model = FastText(
    sentences=sentences,
    vector_size=100,
    window=5,
    min_count=2,
    workers=4,
    epochs=10,
    seed=42
)

print("FastText model trained successfully.")
print("Vocabulary size:", len(fasttext_model.wv))

# ------------------------------------------
# 3. Create one vector for each product
# ------------------------------------------

def create_fasttext_vector(text):
    words = text.split()

    if len(words) == 0:
        return np.zeros(fasttext_model.vector_size)

    word_vectors = [
        fasttext_model.wv[word]
        for word in words
    ]

    return np.mean(word_vectors, axis=0)


fasttext_product_vectors = np.array(
    data["clean_text"].apply(create_fasttext_vector).tolist()
)

print(
    "Product-vector matrix shape:",
    fasttext_product_vectors.shape
)

# ------------------------------------------
# 4. Create the recommendation function
# ------------------------------------------

def recommend_products_fasttext(
    product_name,
    number_of_recommendations=5
):

    # Find products containing the entered text
    matching_products = data[
        data["name"].str.contains(
            product_name,
            case=False,
            na=False,
            regex=False
        )
    ]

    if matching_products.empty:
        print("Product not found. Please try another product name.")
        return

    # Select the first matching product
    product_index = matching_products.index[0]
    selected_product = data.loc[product_index, "name"]

    print(f"\nSelected product: {selected_product}")

    # Calculate cosine similarity
    similarity_scores = cosine_similarity(
        fasttext_product_vectors[product_index].reshape(1, -1),
        fasttext_product_vectors
    ).flatten()

    # Sort products from most similar to least similar
    similar_indices = similarity_scores.argsort()[::-1]

    # Remove the selected product
    similar_indices = [
        index for index in similar_indices
        if index != product_index
    ][:number_of_recommendations]

    # Prepare recommendation results
    result_columns = [
        "name",
        "main_category",
        "sub_category",
        "ratings",
        "discount_price"
    ]

    recommendations = data.loc[
        similar_indices,
        result_columns
    ].copy()

    # Add similarity scores
    recommendations["similarity_score"] = similarity_scores[
        similar_indices
    ]

    # Reset the displayed row numbers
    recommendations.reset_index(drop=True, inplace=True)
    recommendations.index = recommendations.index + 1

    return recommendations

# ------------------------------------------
# 5. Test the recommendation system
# ------------------------------------------

search_product = input("\nEnter a product name: ")

results = recommend_products_fasttext(
    search_product,
    number_of_recommendations=5
)

if results is not None:
    display(results)
```

```python

# Part 8: Compare TF-IDF, Word2Vec and FastText Models

import time
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

# Number of recommendations
k = 5

# Select 50 random products for evaluation
sample_indices = data.sample(
    n=min(50, len(data)),
    random_state=42
).index


# Function to find the top similar products
def get_similar_products(product_index, vectors, k=5):

    # Get the selected product vector
    selected_vector = vectors[product_index]

    # Word2Vec and FastText return a 1D vector.
    # Convert it into a 2D vector for cosine similarity.
    if selected_vector.ndim == 1:
        selected_vector = selected_vector.reshape(1, -1)

    # Calculate similarity scores
    similarity_scores = cosine_similarity(
        selected_vector,
        vectors
    ).flatten()

    # Exclude the selected product itself
    similarity_scores[product_index] = -1

    # Get the top similar products
    similar_indices = similarity_scores.argsort()[::-1][:k]

    return similar_indices


# Function to evaluate one model
def evaluate_model(model_name, vectors):

    main_category_scores = []
    sub_category_scores = []
    execution_times = []

    for product_index in sample_indices:

        start_time = time.time()

        similar_indices = get_similar_products(
            product_index,
            vectors,
            k
        )

        execution_times.append(time.time() - start_time)

        original_main_category = data.loc[
            product_index, "main_category"
        ]

        original_sub_category = data.loc[
            product_index, "sub_category"
        ]

        recommended_main_categories = data.loc[
            similar_indices, "main_category"
        ]

        recommended_sub_categories = data.loc[
            similar_indices, "sub_category"
        ]

        # Calculate Precision@5
        main_precision = (
            recommended_main_categories == original_main_category
        ).mean()

        sub_precision = (
            recommended_sub_categories == original_sub_category
        ).mean()

        main_category_scores.append(main_precision)
        sub_category_scores.append(sub_precision)

    return {
        "Model": model_name,
        "Main Category Precision@5": np.mean(main_category_scores),
        "Subcategory Precision@5": np.mean(sub_category_scores),
        "Average Search Time": np.mean(execution_times)
    }


# Evaluate all three models
tfidf_result = evaluate_model(
    "TF-IDF",
    tfidf_matrix
)

word2vec_result = evaluate_model(
    "Word2Vec",
    product_vectors
)

fasttext_result = evaluate_model(
    "FastText",
    fasttext_product_vectors
)

# Create the comparison table
comparison = pd.DataFrame([
    tfidf_result,
    word2vec_result,
    fasttext_result
])

# Format the results
comparison["Main Category Precision@5"] = (
    comparison["Main Category Precision@5"].round(3)
)

comparison["Subcategory Precision@5"] = (
    comparison["Subcategory Precision@5"].round(3)
)

comparison["Average Search Time"] = (
    comparison["Average Search Time"].round(4)
)

# Display results
print("Model Evaluation Results:")
display(comparison)

# Display the best model
best_model = comparison.loc[
    comparison["Subcategory Precision@5"].idxmax(),
    "Model"
]

print("Best model based on Subcategory Precision@5:", best_model)
```
